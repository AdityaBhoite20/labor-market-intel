```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
```

```python
postings = pd.read_csv("../data/raw/kaggle_job_postings/job_postings.csv")
us_postings = postings[postings["search_country"] == "United States"]

titles = us_postings["job_title"].dropna().unique()
print(len(titles))
print(titles[:20])
```

```python
vectorizer = TfidfVectorizer(stop_words="english", max_features=500)
X_titles = vectorizer.fit_transform(titles)

print(X_titles.shape)
```

```python
kmeans_titles = KMeans(n_clusters=30, random_state=42, n_init=10)
title_clusters = kmeans_titles.fit_predict(X_titles)

title_df = pd.DataFrame({"job_title": titles, "cluster": title_clusters})

print(title_df["cluster"].value_counts().head(10))
```

```python
for cluster_num in [10, 5, 6]:
    print(f"\n=== Cluster {cluster_num} ===")
    sample_titles = title_df[title_df["cluster"] == cluster_num]["job_title"].head(15)
    print(sample_titles.tolist())
```

```python
kmeans_titles_v2 = KMeans(n_clusters=60, random_state=42, n_init=10)
title_clusters_v2 = kmeans_titles_v2.fit_predict(X_titles)

title_df["cluster_v2"] = title_clusters_v2

print(title_df["cluster_v2"].value_counts().head(10))
```

```python
print(title_df[title_df["cluster_v2"] == 39]["job_title"].head(20).tolist())
```

```python
title_df.to_csv("../data/processed/job_title_clusters.csv", index=False)
print("Saved.")
```
